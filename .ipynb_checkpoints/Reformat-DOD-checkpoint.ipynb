```python
import pandas as pd
```

```python
df = pd.read_csv("euthanasia_log.csv", dtype=str)
```

```python
# Parse, handling 2-digit years:
df["DOD_parsed"] = pd.to_datetime(df["DOD"], format="%m/%d/%y", errors="coerce")
```

```python
# Add missing century for ambiguous 2-digit years:
mask_missing = df["DOD_parsed"].isna()
df.loc[mask_missing, "DOD_parsed"] = pd.to_datetime(
    df.loc[mask_missing, "DOD"], format="%m/%d/%Y", errors="coerce"
)
```

```python
# Reformat to the desired string format:
df["DOD"] = df["DOD_parsed"].dt.strftime("%m/%d/%Y")
```

```python
# Drop the helper column and overwrite:
df = df.drop(columns="DOD_parsed")
```

```python
df.to_csv("euthanasia_log.csv", index=False)
```
